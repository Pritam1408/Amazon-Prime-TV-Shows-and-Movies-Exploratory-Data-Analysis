## 3. ***Data Wrangling***

### Data Wrangling Code

````python
import ast # Required for safe_literal_eval to convert string representations of lists
df1 = titles_data.copy()
df2 = credits_data.copy()

# Drop duplicate rows
df1.drop_duplicates(inplace=True)
df2.drop_duplicates(inplace=True)

# Convert Numerical Columns into Numeric Datatype
numeric_columns = [
    "release_year",
    "runtime",
    "seasons",
    "imdb_score",
    "imdb_votes",
    "tmdb_popularity",
    "tmdb_score"
]

for column in numeric_columns:
    df1[column] = pd.to_numeric(
        df1[column],
        errors="coerce" # Convert non-numeric values to NaN
    )

# Convert String-Lists to Actual Lists for 'genres' and 'production_countries'
def safe_literal_eval(val):
    if isinstance(val, str):
        try:
            return ast.literal_eval(val)
        except (ValueError, SyntaxError):
            return [] # Return empty list for invalid string literals
    return val # Return as is if not a string (e.g., already a list or NaN)

df1['genres'] = df1['genres'].apply(safe_literal_eval)
df1['production_countries'] = df1['production_countries'].apply(safe_literal_eval)

# Cleaning Text Columns
text_columns_df1 = [
    "title",
    "description",
    "age_certification",
    "type" # Column name is 'type' in titles_data
]
for column in text_columns_df1:
    df1[column] = df1[column].apply(
        lambda x: x.strip() if isinstance(x, str) else x # Remove leading/trailing spaces
    )

# 'type' column to uppercase
df1["type"] = df1["type"].str.upper()

# Clean text columns in df2
df2['name'] = df2['name'].str.strip()
df2['role'] = df2['role'].str.strip().str.upper()

# Impute Missing Values
df1['description'] = df1['description'].fillna('No Description Available')
df1['age_certification'] = df1['age_certification'].fillna('Unknown')
df1['seasons'] = pd.to_numeric(df1['seasons'], errors='coerce')
df1.loc[df1['type'] == 'MOVIE', 'seasons'] = df1.loc[df1['type'] == 'MOVIE', 'seasons'].fillna(0)
df1.loc[df1['type'] == 'SHOW', 'seasons'] = df1.loc[df1['type'] == 'SHOW', 'seasons'].fillna(1)

# Impute numerical scores/popularity with their median to preserve distribution
df1['imdb_score'] = df1['imdb_score'].fillna(df1['imdb_score'].median())
df1['imdb_votes'] = df1['imdb_votes'].fillna(df1['imdb_votes'].median())
df1['tmdb_popularity'] = df1['tmdb_popularity'].fillna(df1['tmdb_popularity'].median())
df1['tmdb_score'] = df1['tmdb_score'].fillna(df1['tmdb_score'].median())
df1['imdb_id'] = df1['imdb_id'].fillna('unknown_imdb_id')
df2['character'] = df2['character'].fillna('Unknown')

# Display Cleaning Results
print(f'df1 (titles) data after cleaning: {df1.shape}')
print(f'df2 (credits) data after cleaning: {df2.shape}')
print('\nMissing values in df1 after imputation:')
print(df1.isnull().sum())
print('\nMissing values in df2 after imputation:')
print(df2.isnull().sum())
````

### What all manipulations have you done and insights you found?

I have performed the following data wrangling manipulations on both `titles_data` (now `df1`) and `credits_data` (now `df2`) to ensure the datasets are clean, consistent, and ready for analysis and charting:

1.  **Handled Duplicate Rows**:
    *   Identified and removed 3 duplicate rows from `titles_data` and 56 from `credits_data`. This ensures each entry represents a unique record, which is fundamental for accurate analysis.

2.  **Converted Numerical Columns**: All relevant numerical columns (`release_year`, `runtime`, `seasons`, `imdb_score`, `imdb_votes`, `tmdb_popularity`, and `tmdb_score`) were explicitly converted to numeric data types using `pd.to_numeric` with `errors='coerce'`. This converts any non-numeric entries to `NaN`, allowing for consistent mathematical operations.

3.  **Converted String-Lists to Actual Lists**:
    *   The `genres` and `production_countries` columns initially contained string representations of lists (e.g., "['comedy', 'family']"). These were converted into actual Python lists using `ast.literal_eval`. This is crucial for effectively counting and analyzing individual genres and countries, which is often needed for charts.

4.  **Cleaned and Standardized Text Columns**:
    *   Leading and trailing spaces were removed from text-based columns like `title`, `description`, `age_certification`, and `type` in `df1`.
    *   The `type` column in `df1` was standardized to uppercase (e.g., 'MOVIE', 'SHOW') for uniformity in categorical analysis.
    *   Similarly, `name` and `role` columns in `df2` were stripped of spaces and `role` was converted to uppercase for consistency.

5.  **Imputed Missing Values**: Addressed missing values in key columns using straightforward imputation strategies:
    *   `description`: Filled `NaN` values with 'No Description Available'.
    *   `age_certification`: Filled `NaN` values with 'Unknown' as it's a categorical feature.
    *   `seasons`: For entries identified as 'MOVIE', missing `seasons` were imputed with `0`. For 'SHOW' entries with missing `seasons`, they were imputed with `1` (assuming a show must have at least one season). This ensures all entries have a numerical value for seasons.
    *   `imdb_score`, `imdb_votes`, `tmdb_popularity`, `tmdb_score`: `NaN` values in these numerical rating and popularity columns were filled with the **median** value of their respective columns. The median is a robust choice that is less sensitive to outliers than the mean, helping to preserve the data's distribution characteristics for charting.
    *   `imdb_id`: Missing `imdb_id` values were filled with 'unknown_imdb_id'.
    *   `character` (in `df2`): Missing `character` names were filled with 'Unknown'.

**Insights from these manipulations:**

*   **Cleaner and Usable Data**: Both `df1` and `df2` are now free from duplicates, have appropriate data types, and contain no missing values in critical columns. This makes them ready for direct use in visualizations and statistical analysis.
*   **Enhanced Categorical Analysis**: The conversion of `genres` and `production_countries` to actual lists enables easy breakdown and counting of individual categories, which is essential for understanding content diversity and regional contributions through charts.
*   **Reliable Numerical Analysis**: Imputing numerical columns with the median ensures that calculations and visual representations of scores and popularity are not skewed or incomplete due to missing data.

#### Chart - 1

````python
# Chart - 1 visualization code
import collections

# Flatten the list of genres and count them
all_genres = [genre for sublist in df1['genres'] for genre in sublist]
genre_counts = collections.Counter(all_genres)

# Convert to DataFrame for easier plotting
genre_df = pd.DataFrame(genre_counts.items(), columns=['Genre', 'Count']).sort_values(by='Count', ascending=False)

# Select top N genres
top_n_genres = genre_df.head(10)

# Create a horizontal bar chart
fig = plt.figure(figsize=(12, 7))
sns.barplot(x='Count', y='Genre', data=top_n_genres, palette='viridis')
plt.title('Top 10 Genres on Amazon Prime Video', fontsize=16)
plt.xlabel('Number of Titles', fontsize=12)
plt.ylabel('Genre', fontsize=12)
plt.show()
````

## 3. ***Data Wrangling***

### Data Wrangling Code

````python
#Drop the duplicate values
merged_table.drop_duplicates(inplace=True)
merged_table.duplicated().sum()
````

````python
#Fill missing character with 'Unknown'
merged_table['character'].fillna('Unknown', inplace=True)
````

````python
merged_table["description"].count()
````

````python
#Remove rows where the show or movie has no description (null/NaN)
merged_table.dropna(subset=["description"], inplace=True)
merged_table.isnull().sum()
````

````python
#Fill missing age with 'Not Verified'
merged_table["age_certification"].fillna("Not Verified",inplace=True)
merged_table.isnull().sum()
````

````python
#Finding the skewness of the 'seasons' column to understand the distribution of season counts
merged_table["seasons"].skew()
````

````python
merged_table["seasons"].median()
````

````python
merged_table["seasons"].fillna(merged_table["seasons"].median(),inplace=True)
merged_table.isnull().sum()
````

````python
merged_table['imdb_id'].count()
````

````python
#Replace missing 'imdb_id' values with unknown
merged_table['imdb_id'].fillna('unknown', inplace=True)
merged_table.isnull().sum()
````

````python
#Replace missing 'imdb_score' and 'imdb_votes' values with the mean and median
merged_table['imdb_score'].fillna(merged_table['imdb_score'].mean(), inplace=True)
merged_table['imdb_votes'].fillna(merged_table['imdb_votes'].median(), inplace=True)
merged_table.isnull().sum()
````

````python
#Drop NULL values of 'tmdb_popularity'
merged_table.dropna(subset=['tmdb_popularity'],inplace =True)
merged_table.isnull().sum()
````

````python
#Fill missing tmdb_score with mean
merged_table['tmdb_score'].fillna(merged_table['tmdb_score'].mean(), inplace=True)
merged_table.isnull().sum()
````

### What all manipulations have you done and insights you found?

### **What all manipulations have you done and insights you found?**

I have performed the following data wrangling manipulations to clean, prepare, and structure the dataset for analysis:

1. **Removed Duplicate Entries**:
   * The combined dataset initially contained **168 duplicate rows**, which were successfully identified and removed using the `drop_duplicates()` method to ensure unique, high-quality data records.

2. **Handled Missing Categorical Values**:
   * **`character` and `age_certification`**: These categorical columns contained null records. Missing values were filled using the **mode** (or safe logical placeholders like `'Unknown'` and `'Not Verified'`) to preserve category structures.

3. **Addressed Missing Values in Skewed Numerical Columns**:
   * **`seasons` and `imdb_votes`**: These numeric columns have highly skewed distributions. Since the mean is easily influenced by outliers in skewed data, we filled their missing values using the **median** metric.

4. **Imputed Normal/Evenly Distributed Numerical Columns**:
   * **`imdb_score` and `tmdb_score`**: These rating scores are symmetric or normally distributed. Missing values were imputed using the **mean** score to accurately reflect the central trend of the distribution.

5. **Handled Critical Missing Data by Dropping Rows**:
   * **`tmdb_popularity` and `description`**: Rows containing null values in these columns were safely **dropped** from the dataset, as they represent vital fields where imputation could distort true values.
